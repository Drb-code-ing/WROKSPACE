```python
# 实现一个trim函数
# "   hello world   "
def trim(s):
    # return s.strip()
    start = 0
    end = len(s) - 1
    while start < len(s) and s[start] == " ":
        start += 1
    
    while end >= start and s[end] == " ":
        end -= 1
        
    return s[start : end]

trim("   hello world   ")
```
